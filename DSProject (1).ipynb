```python
skills_missing = pd.read_excel(
    "D:\\Intro to data Science\\M&M\\tepsr_sp410_page_spreadsheet.xlsx",
    sheet_name="Sheet 1",
    usecols="A,B,D,F",
    skiprows=42,
    nrows=12,
    header=None,
    names=["Country", 2021, 2023, 2025],
    na_values=":"
)

skills_missing = skills_missing.melt(
    id_vars="Country",
    var_name="Year",
    value_name="Digital_Skills"
)

display(skills_missing.head())
```

```python
columns = ["Country", "Male", "Male_flag", "Female", "Female_flag"]

employment_missing = pd.concat([
    pd.read_excel(
        employment,
        sheet_name=f"Sheet {sheet}",
        usecols="A:E",
        skiprows=38,
        nrows=9,
        header=None,
        names=columns,
        na_values=":"
    ).assign(Year=year)

    for year, sheet in [(2021, 16), (2023, 18), (2025, 20)]
], ignore_index=True)

display(employment_missing.head())
```

```python
extra = skills_missing.merge(
    employment_missing[
        ["Country", "Year", "Female", "Male"]
    ],
    on=["Country", "Year"],
    how="outer"
)

extra = extra.rename(columns={
    "Female": "Female_ICT",
    "Male": "Male_ICT"
})

combined = pd.concat(
    [combined, extra],
    ignore_index=True
)

combined = combined.drop_duplicates(
    subset=["Country", "Year"]
).sort_values(["Year", "Country"]).reset_index(drop=True)

display(combined[combined["Year"] == 2023])
```

```python
print("Total countries:", combined["Country"].nunique())
print("Total rows:", len(combined))

display(
    combined[combined["Country"].isin([
        "Bosnia and Herzegovina",
        "Montenegro"
    ])]
)
```

```python
# Check the number of countries per year
print(combined.groupby("Year")["Country"].nunique())

# Check for missing values
print("\nMissing values:")
print(combined.isna().sum())
```

```python
eu_countries = [
    "Austria", "Belgium", "Bulgaria", "Croatia", "Cyprus",
    "Czechia", "Denmark", "Estonia", "Finland", "France",
    "Germany", "Greece", "Hungary", "Ireland", "Italy",
    "Latvia", "Lithuania", "Luxembourg", "Malta",
    "Netherlands", "Poland", "Portugal", "Romania",
    "Slovakia", "Slovenia", "Spain", "Sweden"
]

eu = combined[combined["Country"].isin(eu_countries)].copy()

print(eu.shape)
display(eu.head())
```

```python
# Calculate average percentages for each year
eu_avg = eu.groupby("Year")[
    ["Digital_Skills", "Female_ICT", "Male_ICT"]
].mean().round(2)

display(eu_avg)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(
    eu_avg.index,
    eu_avg["Digital_Skills"],
    marker="o"
)

plt.title("Average Digital Skills Across EU Countries (2021–2025)")
plt.xlabel("Year")
plt.ylabel("At Least Basic Digital Skills (%)")
plt.xticks([2021, 2023, 2025])
plt.grid(True, alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    eu_avg.index,
    eu_avg["Female_ICT"],
    marker="o",
    label="Female"
)

plt.plot(
    eu_avg.index,
    eu_avg["Male_ICT"],
    marker="o",
    label="Male"
)

plt.title("Gender Representation in ICT Employment (2021–2025)")
plt.xlabel("Year")
plt.ylabel("Share of ICT Specialists (%)")
plt.xticks([2021, 2023, 2025])
plt.ylim(0, 100)
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```

```python
# Compare female ICT representation across selected countries
selected_countries = [
    "Estonia", "Finland", "Germany", "Spain", "Sweden", "Portugal"
]

plt.figure(figsize=(10, 6))

for country in selected_countries:
    country_data = eu[eu["Country"] == country]

    plt.plot(
        country_data["Year"],
        country_data["Female_ICT"],
        marker="o",
        label=country
    )

plt.title("Female ICT Representation by Country (2021–2025)")
plt.xlabel("Year")
plt.ylabel("Female ICT Specialists (%)")
plt.xticks([2021, 2023, 2025])
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```

```python
# Compare the distribution of female ICT representation by year
female_2021 = eu[eu["Year"] == 2021]["Female_ICT"].dropna()
female_2023 = eu[eu["Year"] == 2023]["Female_ICT"].dropna()
female_2025 = eu[eu["Year"] == 2025]["Female_ICT"].dropna()

plt.figure(figsize=(8, 5))

plt.boxplot([female_2021, female_2023, female_2025])

plt.title("Distribution of Female ICT Representation Across EU Countries")
plt.xlabel("Year")
plt.ylabel("Female ICT Specialists (%)")
plt.xticks([1, 2, 3], [2021, 2023, 2025])
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
# Group EU countries into regions
regions = {
    "Nordic": ["Denmark", "Finland", "Sweden"],
    "Baltic": ["Estonia", "Latvia", "Lithuania"],
    "Western Europe": [
        "Austria", "Belgium", "France", "Germany",
        "Ireland", "Luxembourg", "Netherlands"
    ],
    "Southern Europe": [
        "Cyprus", "Greece", "Italy",
        "Malta", "Portugal", "Spain"
    ],
    "Central & Eastern Europe": [
        "Bulgaria", "Croatia", "Czechia", "Hungary",
        "Poland", "Romania", "Slovakia", "Slovenia"
    ]
}

# Assign a region to each country
for region, countries in regions.items():
    eu.loc[eu["Country"].isin(countries), "Region"] = region

# Check our groupings
display(eu[["Country", "Region"]].drop_duplicates().sort_values("Region"))

print("Countries without a region:", eu[eu["Region"].isna()]["Country"].unique())
```

```python
# Calculate regional averages for 2025
regional_2025 = eu[eu["Year"] == 2025].groupby("Region")["Female_ICT"].mean()

# Sort from lowest to highest
regional_2025 = regional_2025.sort_values()

# Plot regional comparison
plt.figure(figsize=(9, 5))

regional_2025.plot(kind="barh")

plt.title("Female ICT Representation by EU Region (2025)")
plt.xlabel("Female ICT Specialists (%)")
plt.ylabel("Region")
plt.xlim(0, 100)
plt.grid(axis="x", alpha=0.3)
```

```python
# Calculate average female ICT representation by region and year
regional_trends = eu.groupby(
    ["Year", "Region"]
)["Female_ICT"].mean().unstack()

# Plot trends
regional_trends.plot(
    kind="line",
    marker="o",
    figsize=(10, 6)
)

plt.title("Female ICT Representation by EU Region (2021–2025)")
plt.xlabel("Year")
plt.ylabel("Female ICT Specialists (%)")
plt.xticks([2021, 2023, 2025])
plt.legend(title="Region", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.grid(True, alpha=0.3)

plt.show()
```
